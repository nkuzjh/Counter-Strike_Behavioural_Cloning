## Alignment

```python
def world_to_radar(x, y, scale=4.4, offset_x=-2476, offset_y=3239, radar_size=1024):
    '''
        "pos_x"     "-2476" // upper left world coordinate
        "pos_y"     "3239"
        "scale"     "4.4"
    '''
    # 转换为雷达坐标（原点在左下）
    px = (x - offset_x) / scale
    py = (y - offset_y) / scale

    # 图像原点在左上，需翻转 Y 轴
    py = radar_size - py

    return int(px), int(py)
```

```python
def world_to_radar_pixel(world_x, world_y, pos_x=-2476, pos_y=3239, scale=4.4, radar_height=1024):
    """
    将 CS2 游戏世界坐标转换为雷达图像素坐标。

    参数:
        world_x, world_y: 游戏世界坐标
        pos_x, pos_y: 雷达图左上角对应的世界坐标（来自 metadata）
        scale: 缩放比例（1 像素 = scale 世界单位）
        radar_height: 雷达图高度（像素），用于 Y 轴翻转

    返回:
        (pixel_x, pixel_y): 雷达图上的像素坐标（原点在左上）
    """
    # X: 从左到右，世界 X 增加 → 像素 X 增加
    pixel_x = (world_x - pos_x) / scale

    # Y: 雷达图左上角是世界最大 Y，所以世界 Y 越小，像素 Y 越大
    # 先计算相对于左上角的 Y 偏移（世界单位）
    delta_y_world = pos_y - world_y  # 注意：pos_y - world_y
    pixel_y = delta_y_world / scale

    return pixel_x, pixel_y
```

```python
map_positions = []
meta_positions = []
for file_idx in range(1,6):
    expert_dust2_npy = np.load(f'data/dataset_dm_expert_dust2_npy_1_to_5/dm_july2021_expert_{file_idx}.npy', allow_pickle=True)
    for i in range(0, 1000):
        # meta_keys = ['localpos1', 'localpos2', 'localpos3', 'height', 'vel_1', 'vel_2', 'vel_3', 'vel_mag', 'viewangle_vert', 'viewangle_xy',]
        meta_x = expert_dust2_npy[i][1]['localpos1']
        meta_y = expert_dust2_npy[i][1]['localpos2']
        meta_positions.append([meta_x, meta_y])
        x, y = world_to_radar_pixel(meta_x, meta_y)
        map_positions.append([x, y])


print(meta_positions[0:5])

print(map_positions[0:5])
```
